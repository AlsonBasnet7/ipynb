```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data_sales.csv")
```

```python
df.head()
```

```python
#filtering the deparment
```

```python
electronics = df[df["Department"] == "Electronics"]

print(electronics)
```

```python
#NOW WE ARE GOING TO USE THE GRAPH 
```

```python
electronics.groupby("Product")["Sales"].sum().plot(kind="bar")

plt.title("Electronics Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
#now filtering sales data greateer than 500
```

```python
high_sales = df[df["Sales"] > 500]

print(high_sales)
```

```python
#lets use the graph to present this information
```

```python
high_sales.plot(
    x="Customer",
    y="Sales",
    kind="bar"
)

plt.title("Customers with Sales > 500")
plt.ylabel("Sales")
plt.show()
```

```python
#lets filter the region now
```

```python
east = df[df["Region"] == "East"]

print(east)
```

```python
#lets use the grapg to present the information
```

```python
east.groupby("Product")["Sales"].sum().plot(kind="pie", autopct="%1.1f%%")

plt.title("East Region Sales Distribution")
plt.ylabel("")
plt.show()
```

```python
#now lets use the multiple conditon and filer some data from the dataset
```

```python
filtered = df[
    (df["Department"] == "Electronics") &
    (df["Sales"] > 300)
]

print(filtered)
```

```python
#lets visualize them 
filtered.plot(
    x="Product",
    y="Sales",
    kind="bar"
)

plt.title("Electronics with Sales > 300")
plt.ylabel("Sales")
plt.show()
```
