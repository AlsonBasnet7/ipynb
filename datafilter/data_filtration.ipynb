# Data Filtration with Pandas
Generating random data and applying various filtering techniques.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

data = {
    'EmployeeID': range(101, 201),
    'Age': np.random.randint(22, 60, 100),
    'Department': np.random.choice(['IT', 'HR', 'Sales', 'Marketing', 'Finance'], 100),
    'Salary': np.random.randint(40000, 150000, 100),
    'PerformanceScore': np.random.uniform(1.0, 5.0, 100).round(1),
    'IsActive': np.random.choice([True, False], 100, p=[0.8, 0.2])
}

df = pd.DataFrame(data)

df.head()
```

## `==` Equal to
Filtering employees who work in the IT department.

```python
it_employees = df[df['Department'] == 'IT']
it_employees.head()
```

## `!=` Not equal to
Filtering employees who do not work in HR.

```python
non_hr_employees = df[df['Department'] != 'HR']
non_hr_employees.head()
```

## `>` Greater than
Filtering employees with Salary strictly greater than 100,000.

```python
high_earners = df[df['Salary'] > 100000]
high_earners.head()
```

## `<` Less than
Filtering employees younger than 30.

```python
young_employees = df[df['Age'] < 30]
young_employees.head()
```

## `>=` Greater than or equal to
Filtering employees with PerformanceScore of 4.0 or above.

```python
high_performers = df[df['PerformanceScore'] >= 4.0]
high_performers.head()
```

## `<=` Less than or equal to
Filtering employees with Salary up to 60,000.

```python
lower_salary = df[df['Salary'] <= 60000]
lower_salary.head()
```

## `&` AND operator
Filtering for IT employees who earn more than 80,000.

```python
it_high_earners = df[(df['Department'] == 'IT') & (df['Salary'] > 80000)]
it_high_earners.head()
```

## `|` OR operator
Filtering for employees in IT or Sales.

```python
it_or_sales = df[(df['Department'] == 'IT') | (df['Department'] == 'Sales')]
it_or_sales.head()
```

## Multiple condition filtering
Filtering for active employees in Finance with a PerformanceScore >= 3.5 and Salary < 120000.

```python
complex_filter = df[
    (df['IsActive'] == True) & 
    (df['Department'] == 'Finance') & 
    (df['PerformanceScore'] >= 3.5) & 
    (df['Salary'] < 120000)
]
complex_filter.head()
```

## Visualizing the filtered data (Putting data in pictures)
Creating visual representations of our filtered dataframes.

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
active_df = df[df['IsActive'] == True]
sns.barplot(data=active_df, x='Department', y='Salary', errorbar=None, palette="viridis")
plt.title('Average Salary by Department (Active Employees)')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.scatterplot(data=high_performers, x='Age', y='Salary', hue='Department', s=100, palette="deep")
plt.title('Age vs Salary (Performance Score >= 4.0)')

plt.tight_layout()
plt.show()
```
