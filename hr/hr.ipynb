```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
salary = (
    df.groupby("JobRole")["MonthlyIncome"]
    .mean()
    .sort_values(ascending=False)
)

salary
```

```python
salary.plot(kind="bar", figsize=(10,5))
plt.title("Average Monthly Income by Job Role")
plt.xticks(rotation=45)
plt.show()
```

```python
dept_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

dept_attrition
```

```python
overtime = pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)

overtime
```

```python
overtime.plot(kind="bar")
plt.title("Overtime vs Attrition")
plt.show()
```

```python
corr = df.select_dtypes(include="number").corr()

corr
```

```python
plt.figure(figsize=(10,8))
plt.imshow(corr)
plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Employees")
plt.show()
```

```python
plt.scatter(
    df["Age"],
    df["MonthlyIncome"]
)

plt.title("Age vs Monthly Income")
plt.xlabel("Age")
plt.ylabel("Monthly Income")

plt.show()
```

```python
df["Gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Gender Distribution")
plt.show()
```

```python
top_salary = df.nlargest(
    10,
    "MonthlyIncome"
)

top_salary[
    [
        "JobRole",
        "Department",
        "Age",
        "MonthlyIncome"
    ]
]
```

```python
bins = [18, 25, 35, 45, 55, 65]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)

df[["Age", "AgeGroup"]].head()
```

```python
age_attrition = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"]
)

age_attrition
```

```python
age_attrition.plot(kind="bar")
plt.title("Age Group vs Attrition")
plt.show()
```
