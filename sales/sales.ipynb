```python
# This program performs a simple sales data analysis using the Pandas library.
# It loads a sales dataset from a CSV file, removes any missing values,
# and calculates important business metrics such as revenue, profit,
# and profit margin for each product. The program also filters high-selling
# products, summarizes profit by region, identifies the products with the
# highest profit margins, calculates the average revenue for each product
# category, and computes the correlation between revenue, profit, and
# quantity sold to better understand the relationships within the data.
```

```python

import pandas as pd
```

```python
df = pd.read_csv("sales_data.csv")
```

```python
df.head()
```

```python
df = df.dropna()
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
df["Profit"] = df["Revenue"] - df["Cost"]
```

```python
high_sales = df[df["Revenue"] > 1000]
region_profit = high_sales.groupby("Region")["Profit"].sum()
```

```python
df["Margin"] = (df["Profit"] / df["Revenue"]) * 100
top = df.sort_values("Margin", ascending=False)
```

```python
print(region_profit)
print(top[["Product", "Region", "Revenue", "Profit"]].head())
```

```python
avg = df.groupby("Category")["Revenue"].mean()
print(avg)
```

```python
corr = df[["Revenue", "Profit", "Quantity"]].corr()
print(corr)
```
